### In-Context Few-Shot Prompt & Response Audit
Inspect prompt logs, raw model completions, and parsing accuracy from Ollama LLaMA-3.1 8B.


```python
exp_dir = Path.cwd().parent / "experiments" / "Foundation" / "LLaMA_3.1_8B_Ollama"
if exp_dir.exists():
    runs = list(exp_dir.iterdir())
    if runs:
        log_file = runs[0] / "llm_prompts_and_raw_responses.jsonl"
        if log_file.exists():
            records = [json.loads(line) for line in open(log_file, encoding='utf-8')]
            print(f"Total prompt completions logged: {len(records)}")
            print("\nSample Prompt & Response:")
            sample = records[0]
            print("Target:", sample["target_crop"], sample["target_year"])
            print("Raw Response:", repr(sample["raw_response"]))
            print("Parsed Cost:", sample["parsed_cost"], f"LKR (Parsed: {sample['parse_success']})")
```
