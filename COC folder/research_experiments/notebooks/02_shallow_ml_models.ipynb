### Visual Comparison: Shallow ML Performance
Ranking classical ML models by Leave-One-Season-Year-Out MAE.


```python
import matplotlib.pyplot as plt
import seaborn as sns

if results_csv.exists():
    df_lead = pd.read_csv(results_csv)
    ml_lead = df_lead[df_lead["family"] == "Shallow_ML"].sort_values(by="LOSO_MAE")
    plt.figure(figsize=(10, 6))
    sns.barplot(data=ml_lead, y="model_name", x="LOSO_MAE", palette="viridis")
    plt.title("Family B Leaderboard (LOSO MAE)", fontweight="bold")
    plt.xlabel("MAE (LKR / acre)", fontweight="bold")
    plt.ylabel("Model", fontweight="bold")
    plt.tight_layout()
    plt.show()
```
